# População brasileira: 2010 e 2022

Agregações por estado e município, ordenadas pela variação absoluta da população entre 2010 e 2022.

```python
import pandas as pd

ARQUIVO_ENTRADA = 'CD2022_Populacao_2010_Compatibilizada_20231222.xlsx'

df = pd.read_excel(ARQUIVO_ENTRADA, sheet_name=0, skiprows=2)
df = df[df['COD. UF'].notna()].copy()
df.columns = ['IGNORAR', 'UF', 'COD_UF', 'COD_MUNIC', 'MUNICIPIO', 'POP_2010_SINOPSE', 'POP_2010_COMPATIBILIZADA', 'POP_2022']

col_2010 = 'POP_2010_COMPATIBILIZADA'
col_2022 = 'POP_2022'
df['Crescimento 2010-2022'] = df[col_2022] - df[col_2010]

df.head()
```

```python
populacao_estado = (
    df.groupby('UF')[[col_2010, col_2022, 'Crescimento 2010-2022']]
      .sum()
      .reset_index()
      .sort_values('Crescimento 2010-2022', ascending=False)
)

populacao_estado.to_csv('populacao_por_estado.csv', sep=';', index=False, encoding='utf-8-sig')
populacao_estado
```

```python
populacao_municipio = (
    df[['UF', 'COD_UF', 'COD_MUNIC', 'MUNICIPIO', col_2010, col_2022, 'Crescimento 2010-2022']]
      .sort_values('Crescimento 2010-2022', ascending=False)
)

populacao_municipio.to_csv('populacao_por_municipio.csv', sep=';', index=False, encoding='utf-8-sig')
populacao_municipio.head()
```
